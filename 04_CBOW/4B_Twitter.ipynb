```python
import pandas as pd
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
```

```python
tweets = pd.read_csv("twitter_dataset.csv")["Text"].dropna()
sentences = tweets.apply(simple_preprocess)
sentences = sentences[sentences.str.len() > 0].tolist()
```

```python
cbow = Word2Vec(sentences, vector_size=100, min_count=1, sg=0, epochs=20, workers=1, seed=42)
```

```python
word = cbow.wv.index_to_key[0]
print("Word:", word, "\n", cbow.wv.most_similar(word, topn=5))
```
