```python
from pathlib import Path
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
```

```python
text = Path("document3.txt").read_text(encoding="utf-8")
sentences = [simple_preprocess(sentence) for sentence in text.split(".")]
```

```python
cbow = Word2Vec(sentences, vector_size=100, min_count=1, sg=0, epochs=100, workers=1, seed=42)
```

```python
word = cbow.wv.index_to_key[0]
print("Word:", word, "\n", cbow.wv.most_similar(word, topn=5))
```
