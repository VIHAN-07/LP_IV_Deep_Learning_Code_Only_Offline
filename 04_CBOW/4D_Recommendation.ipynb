```python
import pandas as pd
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tweets = pd.read_csv("twitter_dataset.csv").dropna(subset=["Text"])
tweets["words"] = tweets["Text"].apply(simple_preprocess)
tweets = tweets[tweets["words"].str.len() > 0].copy()
sentences = tweets["words"].tolist()
```

```python
cbow = Word2Vec(sentences, vector_size=100, min_count=1, sg=0, epochs=20, workers=1, seed=42)
```

```python
vectors = [cbow.wv.get_mean_vector(words) for words in sentences]
```

```python
query = input("Enter keywords from the dataset: ")
query_vector = cbow.wv.get_mean_vector(simple_preprocess(query))
tweets["similarity"] = cosine_similarity([query_vector], vectors)[0]
results = tweets.sort_values("similarity", ascending=False).head(5)
print(results[["Text", "similarity"]].to_string(index=False))
```
