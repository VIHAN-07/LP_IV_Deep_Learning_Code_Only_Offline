```python
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.optimizers import SGD
```

```python
train = pd.read_csv("mnist_train.csv", dtype="uint8")
test = pd.read_csv("mnist_test.csv", dtype="uint8")
y_train, y_test = train.pop("label").to_numpy(), test.pop("label").to_numpy()
x_train = train.to_numpy("float32") / 255
x_test = test.to_numpy("float32") / 255
```

```python
model = Sequential([
    Input((784,)), Dense(256, activation="relu"), Dense(128, activation="relu"),
    Dropout(0.2), Dense(10, activation="softmax")])
model.compile(optimizer=SGD(learning_rate=0.1, momentum=0.9), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
history = model.fit(x_train, y_train, epochs=11, batch_size=128, validation_split=0.1)
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print("Test accuracy (%):", round(100 * accuracy, 2))
```

```python
pd.DataFrame(history.history).plot(y=["loss", "accuracy"], subplots=True)
plt.show()
```
