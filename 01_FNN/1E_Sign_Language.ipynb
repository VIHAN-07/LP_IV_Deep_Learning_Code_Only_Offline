```python
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.preprocessing import LabelEncoder
```

```python
train = pd.read_csv("sign_mnist_train.csv", dtype="uint8")
test = pd.read_csv("sign_mnist_test.csv", dtype="uint8")
y_train, y_test = train.pop("label"), test.pop("label")
x_train = train.to_numpy("float32") / 255
x_test = test.to_numpy("float32") / 255
encoder = LabelEncoder()
y_train, y_test = encoder.fit_transform(y_train), encoder.transform(y_test)
```

```python
model = Sequential([Input((784,)), Dense(128, "relu"), Dense(24, "softmax")])
model.compile("adam", "sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
h = model.fit(x_train, y_train, epochs=20, batch_size=128)
```

```python
print(model.evaluate(x_test, y_test))
```

```python
pd.DataFrame(h.history).plot(subplots=True)
plt.show()
```
