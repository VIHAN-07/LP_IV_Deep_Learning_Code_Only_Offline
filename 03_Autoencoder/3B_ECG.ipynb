```python
import numpy as np
import pandas as pd
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("ecg.csv", header=None, dtype="float32")
train, test = train_test_split(df, test_size=0.2, stratify=df[140], random_state=42)
scaler = MinMaxScaler()
x_train = scaler.fit_transform(train[train[140] == 1].drop(columns=140))
x_test = scaler.transform(test.drop(columns=140))
```

```python
model = Sequential([Input((140,)), Dense(16, "relu"), Dense(140, "sigmoid")])
model.compile(optimizer="adam", loss="mse")
model.fit(x_train, x_train, epochs=10, batch_size=128)
```

```python
error = np.mean((x_train - model.predict(x_train, verbose=0)) ** 2, axis=1)
threshold = np.percentile(error, 95)
error = np.mean((x_test - model.predict(x_test, verbose=0)) ** 2, axis=1)
anomalies = error > threshold
print("Anomalies:", anomalies.sum())
```
